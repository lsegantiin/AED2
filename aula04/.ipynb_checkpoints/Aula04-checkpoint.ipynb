```python
livros = [
    ("A Hora da Estrela", "Clarice Lispector", 1977),
    ("Capitães da Areia", "Jorge Amado", 1937),
    ("Dom Casmurro", "Machado de Assis", 1899),
    ("Grande Sertão: Veredas", "João Guimarães Rosa", 1956),
    ("Iracema", "José de Alencar", 1865),
    ("Memórias Póstumas de Brás Cubas", "Machado de Assis", 1881),
    ("O Alienista", "Machado de Assis", 1882),
    ("O Cortiço", "Aluísio Azevedo", 1890),
    ("Senhora", "José de Alencar", 1875),
    ("Vidas Secas", "Graciliano Ramos", 1938)
]
```

```python
def busca_binaria(livros, titulo):
    inicio = 0
    fim = len(livros) - 1

    while inicio <= fim:
        meio = (inicio + fim) // 2

        titulo_meio = livros[meio][0]

        if titulo_meio.lower() == titulo.lower():
            return livros[meio]

        elif titulo.lower() < titulo_meio.lower():
            fim = meio - 1

        else:
            inicio = meio + 1

    return None
```

```python
titulo = input("Digite o título do livro: ")

resultado = busca_binaria(livros, titulo)

if resultado:
    print("\nLivro encontrado!")
    print("Título:", resultado[0])
    print("Autor:", resultado[1])
    print("Ano:", resultado[2])
else:
    print("\nLivro não encontrado.")
```
