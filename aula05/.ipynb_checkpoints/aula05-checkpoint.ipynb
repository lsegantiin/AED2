```python
##Parte B — Programação: nível fácil
# Exercício 4 — Classe Livro: atributos e métodos 

class Livro:
    def __init__(self, titulo, autor, paginas):
        self.titulo = titulo
        self.autor = autor
        self.paginas = paginas

    def descricao(self):
        return f"{self.titulo} - {self.autor} ({self.paginas} páginas)"

livro1 = Livro("Dom Casmurro", "Machado de Assis", 256)
livro2 = Livro("1984", "George Orwell", 328)

print(livro1.descricao())
print(livro2.descricao())
```

```python
# Exercício 5 — Conta simples com encapsulamento 

class Conta:
    def __init__(self, titular):
        self.titular = titular
        self.__saldo = 0

    def depositar(self, valor):
        if valor > 0:
            self.__saldo += valor

    def consultar_saldo(self):
        return self.__saldo

conta = Conta("Laura")
conta.depositar(100)
conta.depositar(50)
print(conta.consultar_saldo())
```

```python
# Exercício 6 — Animais: herança e polimorfismo

class Animal:
    def __init__(self, nome):
        self.nome = nome

    def emitir_som(self):
        return "som genérico"

class Cachorro(Animal):
    def emitir_som(self):
        return "au au"

class Gato(Animal):
    def emitir_som(self):
        return "miau"

animais = [Cachorro("leleco"), Gato("Chico")]
for animal in animais:
    print(animal.emitir_som())
```

```python
## Parte C — Programação: nível médio
# Exercício 7 — Veículo com velocidade controlada

class Veiculo:
    def __init__(self, placa, velocidade_max):
        self.__placa = placa
        self.__velocidade_max = velocidade_max
        self.__velocidade_atual = 0

    def get_placa(self):
        return self.__placa

    def get_velocidade_max(self):
        return self.__velocidade_max

    def set_velocidade_max(self, nova_max):
        if nova_max > 0 and nova_max >= self.__velocidade_atual:
            self.__velocidade_max = nova_max

    def acelerar(self, incremento=10):
        self.__velocidade_atual = min(self.__velocidade_atual + incremento, self.__velocidade_max)

    def frear(self, decremento=10):
        self.__velocidade_atual = max(self.__velocidade_atual - decremento, 0)

    def __str__(self):
        return f"Placa: {self.__placa}, Velocidade atual: {self.__velocidade_atual}, Máxima: {self.__velocidade_max}"

carro = Veiculo("EPF-0326", 100)
carro.acelerar()
carro.acelerar(50)
carro.frear(20)
print(carro)
```

```python
# Exercício 8 — Funcionários e cálculo de bônus

class Funcionario:
    def __init__(self, nome, salario):
        self.nome = nome
        self.salario = salario

    def calcular_bonus(self):
        return self.salario * 0.05

    def __str__(self):
        return f"{self.nome} - Salário: {self.salario}"

class Vendedor(Funcionario):
    def __init__(self, nome, salario, vendas):
        super().__init__(nome, salario)
        self.vendas = vendas

    def calcular_bonus(self):
        return self.salario * 0.05 + self.vendas * 0.02

class Gerente(Funcionario):
    def __init__(self, nome, salario, equipe):
        super().__init__(nome, salario)
        self.equipe = equipe

    def calcular_bonus(self):
        return self.salario * 0.10 + 100 * len(self.equipe)

funcionarios = [
    Funcionario("Ana", 3000),
    Vendedor("Carlos", 2000, 5000),
    Gerente("Mariana", 5000, ["Ana", "Carlos"])
]

for f in funcionarios:
    print(f"{f.nome} ({f.__class__.__name__}) - Bônus: {f.calcular_bonus()}")
```

```python
# Exercício 9 — Produtos e contador de objetos

class Produto:
    contador = 0

    def __init__(self, nome, preco):
        Produto.contador += 1
        self.__id = Produto.contador
        self.__nome = nome
        self.__preco = preco

    def get_id(self):
        return self.__id

    def get_nome(self):
        return self.__nome

    def get_preco(self):
        return self.__preco

    def set_preco(self, novo_preco):
        if novo_preco > 0:
            self.__preco = novo_preco

    def aplicar_desconto(self, percentual):
        if 0 <= percentual <= 100:
            self.__preco *= (1 - percentual/100)

    def __str__(self):
        return f"Produto {self.__id}: {self.__nome} - R${self.__preco:.2f}"

p1 = Produto("Notebook", 3000)
p2 = Produto("Celular", 1500)
p3 = Produto("Fone", 200)

print(p1, p2, p3, sep="\n")
print("Total de produtos criados:", Produto.contador)
```

```python
## Parte D — Desafio integrador

#Exercício 10 — Mini sistema de pedidos de e-commerce

class Produto:
    contador = 0
    def __init__(self, nome, preco):
        Produto.contador += 1
        self.__id = Produto.contador
        self.__nome = nome
        self.__preco = preco

    def get_preco(self):
        return self.__preco

    def calcular_frete(self):
        return 0

    def __str__(self):
        return f"{self.__nome} - R${self.__preco:.2f}"

class ProdutoFisico(Produto):
    def __init__(self, nome, preco, peso):
        super().__init__(nome, preco)
        self.peso = peso

    def calcular_frete(self):
        return 5 + 2 * self.peso

class ProdutoDigital(Produto):
    def __init__(self, nome, preco, tamanho_mb):
        super().__init__(nome, preco)
        self.tamanho_mb = tamanho_mb

class Cliente:
    def __init__(self, nome, email):
        self.__nome = nome
        self.__email = email

    def __str__(self):
        return f"{self.__nome} ({self.__email})"

class Pedido:
    contador = 0
    def __init__(self, cliente):
        Pedido.contador += 1
        self.__numero = Pedido.contador
        self.cliente = cliente
        self.produtos = []
        self.__status = "aberto"

    def adicionar_produto(self, produto):
        if self.__status == "aberto":
            self.produtos.append(produto)

    def calcular_total(self):
        return sum(p.get_preco() for p in self.produtos)

    def calcular_frete_total(self):
        return sum(p.calcular_frete() for p in self.produtos)

    def fechar_pedido(self):
        self.__status = "fechado"

    def __str__(self):
        return f"Pedido {self.__numero} - Cliente: {self.cliente}\nTotal: R${self.calcular_total():.2f}\nFrete: R${self.calcular_frete_total():.2f}"

cliente = Cliente("Laura", "laura.segantin@gmail.com")
p1 = ProdutoFisico("Notebook", 3000, 2)
p2 = ProdutoDigital("Curso Python", 200, 500)
p3 = ProdutoFisico("Livro", 100, 1)

pedido = Pedido(cliente)
pedido.adicionar_produto(p1)
pedido.adicionar_produto(p2)
pedido.adicionar_produto(p3)

print(pedido)
pedido.fechar_pedido()
pedido.adicionar_produto(ProdutoDigital("Extra", 50, 100))  # não será adicionado

```
