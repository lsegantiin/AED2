```python
# Acesso ao primeiro elemento

def imprime_primeiro(lista):
    print(lista[0])
```

```python
Complexidade: O(1) — constante
```

```python
#Encontrando o maior valor

def encontrar_maximo(lista):
    valor_maximo = lista[0]

    for numero in lista:
        if numero > valor_maximo:
            valor_maximo = numero

    return valor_maximo
```

```python
Complexidade de tempo: O(n) — linear.
```

```python
#Complexidade da busca binária

def busca_binaria(lista, alvo):
    esquerda = 0
    direita = len(lista) - 1

    while esquerda <= direita:
        meio = (esquerda + direita) // 2

        if lista[meio] == alvo:
            return meio
        elif lista[meio] < alvo:
            esquerda = meio + 1
        else:
            direita = meio - 1

    return -1
```

```python
Complexidade: O(log n) — logarítmica
```

```python
#Duas formas de localizar as menores idades

def existe_2_menores_v1(idades):
    tamanho = len(idades)
    menor = 200

    for i in range(tamanho):
        if idades[i] < menor:
            menor = idades[i]

    cont = 0

    for i in range(tamanho):
        if idades[i] == menor:
            cont += 1

    return cont > 1


def existe_2_menores_v2(idades):
    idades.sort()
    return idades[0] == idades[1]
```

```python
v1: O(n)
v2: O(n log n)
Mais eficiente assintoticamente: v1
```

```python
# Fibonacci recursivo

def fib(n):
    if n < 2:
        return n
    else:
        return fib(n - 1) + fib(n - 2)
```

```python
Complexidade: O(2^n) — exponencial
```

```python
#Busca linear ou ordenar e buscar?

def busca_linear(lista, alvo):
    for elemento in lista:
        if elemento == alvo:
            return True

    return False
```

```python
#Análise de Desempenho de Busca em Dados de Clientes

def buscar_aluno_linear(lista_alunos, cpf):
    for aluno in lista_alunos:
        if aluno["cpf"] == cpf:
            return aluno
    return None


def buscar_aluno_eficiente(dicionario_alunos, cpf):
    if cpf in dicionario_alunos:
        return dicionario_alunos[cpf]
    return None


# Exemplo de dados para testar a solução
lista_de_alunos = [
    {"nome": "Lucas Oliveira", "cpf": "145.782.369-10", "curso": "CDN"},
    {"nome": "Mariana Costa", "cpf": "268.451.703-25", "curso": "ADS"},
    {"nome": "Rafael Santos", "cpf": "379.624.815-40", "curso": "CDN"},
    {"nome": "Juliana Almeida", "cpf": "482.936.157-63", "curso": "ADS"},
    {"nome": "Gabriel Ferreira", "cpf": "593.147.268-71", "curso": "CDN"}
]


# Pré-processamento: cria um índice cuja chave é o CPF
dicionario_de_alunos = {
    aluno["cpf"]: aluno for aluno in lista_de_alunos
}


cpf_buscado = "379.624.815-40"

print(buscar_aluno_linear(lista_de_alunos, cpf_buscado))
print(buscar_aluno_eficiente(dicionario_de_alunos, cpf_buscado))
```

```python
# Recomendação de Disciplinas Baseada em Interesses 

def recomendar_disciplinas(historicos_alunos, historico_aluno_principal):
    recomendacoes = set()

    for historico_outro_aluno in historicos_alunos:
        if historico_outro_aluno != historico_aluno_principal:
            
            intersecao = (
                set(historico_aluno_principal)
                & set(historico_outro_aluno)
            )

            if intersecao:
                diferenca = (
                    set(historico_outro_aluno)
                    - set(historico_aluno_principal)
                )

                recomendacoes.update(diferenca)

    return list(recomendacoes)


historicos = [
    [
        "Disciplinas, Análise e Estrutura de Dados",
        "Lógica Computacional",
        "Estatística"
    ],
    [
        "Lógica Computacional",
        "Estatística"
    ],
    [
        "Disciplinas, Análise e Estrutura de Dados",
        "Estatística",
        "Banco de Dados"
    ],
    [
        "Programação",
        "Banco de Dados"
    ]
]

historico_principal = [
    "Disciplinas, Análise e Estrutura de Dados",
    "Lógica Computacional"
]

recomendadas = recomendar_disciplinas(
    historicos,
    historico_principal
)

print("Disciplinas recomendadas:", recomendadas)
```
