```python
# A Agenda de Contatos

contatos = {}


def adicionar_contato(nome, telefone, email):
    contatos[nome] = {
        "telefone": telefone,
        "email": email
    }
    print(f"✓ Contato {nome} adicionado com sucesso!\n")


def buscar_contato(nome):
    if nome in contatos:
        contato = contatos[nome]

        print("=" * 40)
        print(f"       INFORMAÇÕES DE {nome.upper()}")
        print("=" * 40)
        print(f"Telefone : {contato['telefone']}")
        print(f"E-mail   : {contato['email']}")
        print("=" * 40)
        print()

    else:
        print("Contato não encontrado.\n")


def listar_contatos():
    print("=" * 50)
    print("              LISTA DE CONTATOS")
    print("=" * 50)

    for nome, contato in contatos.items():
        print(f"Nome     : {nome}")
        print(f"Telefone : {contato['telefone']}")
        print(f"E-mail   : {contato['email']}")
        print("-" * 50)


# Adicionando contatos
adicionar_contato(
    "Vinicius",
    "(15) 1234-5678",
    "vinicius@email.com"
)

adicionar_contato(
    "Mariana",
    "(15) 9876-5432",
    "mariana@email.com"
)


# Buscando um contato
buscar_contato("Vinicius")


# Listando todos os contatos
listar_contatos()
```

```python
# A Lista de Tarefas

tarefas = [
    "Alimentar o gato",
    "Trocar a água do gato",
    "Limpar a areia do gato"
]


def adicionar_tarefa(tarefa):
    tarefas.append(tarefa)
    print(f"Tarefa '{tarefa}' adicionada.")


def marcar_concluida(tarefa):
    if tarefa in tarefas:
        tarefas.remove(tarefa)
        print(f"Tarefa '{tarefa}' marcada como concluída.")
    else:
        print(f"Tarefa '{tarefa}' não encontrada na lista.")


def visualizar_tarefas():
    print("\nLista de tarefas pendentes:")
    for tarefa in tarefas:
        print(f"- {tarefa}")


# Exemplo de uso
adicionar_tarefa("Brincar com o gato")
marcar_concluida("Trocar a água do gato")
visualizar_tarefas()
```
