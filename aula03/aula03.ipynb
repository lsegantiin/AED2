```python
#Exercicio 4

def buscar_cliente(nome, lista_clientes):
    for i in range(len(lista_clientes)):
        if lista_clientes[i] == nome:
            return i
    return -1


# Exemplo de utilização
clientes = ["Ana", "Carlos", "Mariana", "João", "Pedro"]

resultado = buscar_cliente("Mariana", clientes)

if resultado != -1:
    print(f"Cliente encontrado na posição {resultado}.")
else:
    print("Cliente não encontrado.")


resultado = buscar_cliente("Fernanda", clientes)

if resultado != -1:
    print(f"Cliente encontrado na posição {resultado}.")
else:
    print("Cliente não encontrado.")
```

```python
#Exercicio 5

def conta_ocorrencias(valor, lista):
    contador = 0

    for elemento in lista:
        if elemento == valor:
            contador += 1

    return contador


# Casos de teste
print(conta_ocorrencias(3, [1, 3, 5, 3, 7]))
print(conta_ocorrencias(2, [2, 2, 4, 6, 2]))
print(conta_ocorrencias(10, [1, 2, 3, 4, 5]))
print(conta_ocorrencias("Ana", ["João", "Ana", "Pedro", "Ana"]))
```

```python
#Exercicio 6

def remove_elemento(lista, valor):
    for i in range(len(lista)):
        if lista[i] == valor:
            return lista.pop(i)

    return None


# Exemplo
numeros = [10, 20, 30, 20, 40]

removido = remove_elemento(numeros, 20)

print("Valor removido:", removido)
print("Lista após a remoção:", numeros)

# Tentando remover um valor que não existe
removido = remove_elemento(numeros, 100)

print("Valor removido:", removido)
print("Lista após a tentativa:", numeros)
```

```python
#Encontrando seu Filme Favorito

def buscar_filme(titulo, filmes):
    for i in range(len(filmes)):
        if filmes[i].lower() == titulo.lower():
            return i

    return -1


filmes = [
    "O Senhor dos Anéis",
    "Interestelar",
    "Matrix",
    "Titanic",
    "Toy Story",
    "Vingadores"
]

titulo = input("Digite o título do filme que deseja buscar: ")

posicao = buscar_filme(titulo, filmes)

if posicao != -1:
    print(f'O filme "{filmes[posicao]}" foi encontrado!')
    print(f"Ele está na posição {posicao} do catálogo.")
else:
    print(f'O filme "{titulo}" não está no catálogo.')
```

```python
#Verificando a Lista de Convidados

def verificar_convidado(nome, convidados):
    for convidado in convidados:
        if convidado.lower() == nome.lower():
            return True

    return False


convidados = [
    "Ana",
    "Carlos",
    "Mariana",
    "João",
    "Pedro"
]

nome = input("Digite o nome do convidado: ")

if verificar_convidado(nome, convidados):
    print(f"{nome} está na lista de convidados!")
else:
    print(f"{nome} não está na lista de convidados.")
```
