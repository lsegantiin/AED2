```python
#Algoritimo de Busca Sequencial

def busca_seq(l, n):
    for i in range(len(l)):
        if l[i] == n:
            return i
    return -1
```

```python
lista = []

lista.append(5)
lista.append(10)
lista.append(20)
lista.append(21)
lista.append(3)
lista.append(15)
lista.append(1)
lista.append(2)
lista.append(18)
lista.append(26)
```

```python
numero = int(input("Digite o valor a ser pesquisado:"))

indice = busca_seq(lista,numero)

if indice == -1:
    Print(f"O numero {numero} não foi encontrado na lista!")
else:
    print(f"O numero {numero} foi encontrado na posição {indice}")
    
```
