```python
# Percorre uma árvore binária em pré-ordem
def preorder(self):
    # Imprime o valor da raiz
    print(self.key)

    # Visita a subárvore à esquerda
    if self.left_child:
        self.left_child.preorder()

    # Visita a subárvore à direita
    if self.right_child:
        self.right_child.preorder()
```

```python
# Percorre uma árvore binária em ordem
def inorder(self):
    # Visita a subárvore à esquerda
    if self.left_child:
        self.left_child.inorder()

    # Imprime o valor da raiz
    print(self.key)

    # Visita a subárvore à direita
    if self.right_child:
        self.right_child.inorder()
```

```python
# Percorre uma árvore binária em pós-ordem
def postorder(self):
    # Visita a subárvore à esquerda
    if self.left_child:
        self.left_child.postorder()

    # Visita a subárvore à direita
    if self.right_child:
        self.right_child.postorder()

    # Imprime o valor da raiz
    print(self.key)
```
